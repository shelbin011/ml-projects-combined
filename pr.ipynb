```
df = pd.read_csv("world_population.csv")
```

```
india = df[df["Country/Territory"] == "India"]

india
```

```
data = {
    "Year":[1970,1980,1990,2000,2010,2015,2020,2022],

    "Population":[
        india["1970 Population"].values[0],
        india["1980 Population"].values[0],
        india["1990 Population"].values[0],
        india["2000 Population"].values[0],
        india["2010 Population"].values[0],
        india["2015 Population"].values[0],
        india["2020 Population"].values[0],
        india["2022 Population"].values[0]
    ]
}

new_df = pd.DataFrame(data)

new_df
```

```
import matplotlib.pyplot as plt

plt.scatter(new_df["Year"], new_df["Population"])

plt.xlabel("Year")
plt.ylabel("Population")

plt.title("India Population Growth")

plt.show()
```

```
X = new_df[["Year"]]
y = new_df["Population"]
```

```
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=3)

X_poly = poly.fit_transform(X)
```

```
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_poly, y)
```

```
import numpy as np

year = np.array([[2030]])

year_poly = poly.transform(year)

prediction = model.predict(year_poly)

print("Predicted Population in 2030:", int(prediction[0]))
```

```
plt.scatter(X, y)

plt.plot(X, model.predict(X_poly))

plt.xlabel("Year")
plt.ylabel("Population")

plt.title("Population Prediction using Polynomial Regression")

plt.show()
```

```
future_years = [2030, 2040, 2050]

for year in future_years:

    year_array = np.array([[year]])

    year_poly = poly.transform(year_array)

    pred = model.predict(year_poly)

    print("Predicted Population in",year,":", int(pred[0]))
```
